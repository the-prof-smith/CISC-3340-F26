3. Pick two more two-character hex codes of your own, predict their decimal value first, then verify in code.

**Your hand-worked reasoning for A3 (double-click this cell to edit):**



Hex is base 16, so each digit is a power of 16. For a two-digit hex number `XY`:

value = (X × 16¹) + (Y × 16⁰)

For `A3`: A = 10 in decimal, 3 = 3.

(10 × 16) + (3 × 1) = 160 + 3 = **163**

````python
# Part 2: Hex to decimal conversion

# convert "A3" from hex to decimal using int()
member_id = int("A3", 16)
print("A3 in decimal is", member_id)

# pick two more hex codes of your own and convert them too
# Example predictions: "1F" -> (1*16)+15 = 31, "B0" -> (11*16)+0 = 176
code_2 = int("1F", 16)
code_3 = int("B0", 16)
print("1F in decimal is", code_2)
print("B0 in decimal is", code_3)
````

## Part 3: Plan the pricing logic in pseudocode

Nova Cinema's actual pricing rules:

| Rule | Price / Adjustment |
|---|---|
| Child (under 12) | \$8.00 base |
| Senior (65 or older) | \$9.00 base |
| Adult (everyone else) | \$12.50 base |
| Matinee showing (before 5pm) | \$2.00 off the base price |
| Member | an additional \$1.00 off |

Before writing any Python, plan the decision logic in plain English below. Work through: what do you check first? What order do the discounts get applied in? What happens if someone is both a senior and a member at a matinee?

**Your pseudocode (double-click this cell to edit):**



1. Get the customer's age, and whether the showing is a matinee and whether they are a member.
2. Determine the BASE PRICE by checking age: if age is under 12, base price is \$8.00. Otherwise, if age is 65 or older, base price is \$9.00. Otherwise (everyone else), base price is \$12.50. (Child must be checked before senior/adult, since order matters in an if/elif chain — checking child first avoids a 10-year-old accidentally falling into the "everyone else" bucket.)
3. If the showing is a matinee, subtract \$2.00 from the base price.
4. If the customer is a member, subtract an additional \$1.00 from whatever the price is at this point.
5. Print/return the final price.

A senior who is also a member at a matinee gets ALL THREE adjustments applied in sequence: \$9.00 base − \$2.00 (matinee) − \$1.00 (member) = \$6.00. The two discounts are independent of each other and both just stack on top of whatever base price was already chosen.

## Part 4: Implement it with conditionals and Booleans

Turn your Part 3 pseudocode into real Python using `if` / `elif` / `else` and Boolean variables (for example `is_matinee`, `is_member`).

Set your `age`, `is_matinee`, and `is_member` variables by hand at the top of the cell for each test run. Test your program against every row below and confirm your output matches exactly, then change the variables and re-run for the next row.

| Age | Matinee? | Member? | Expected Price |
|---|---|---|---|
| 8 | No | No | \$8.00 |
| 30 | Yes | No | \$10.50 |
| 70 | No | Yes | \$8.00 |
| 30 | Yes | Yes | \$9.50 |

````python
# Part 4: Nova Cinema pricing logic
# Change these three values to test each row in the table, then re-run this cell
age = 30          # try: 8, 30, 70, 30
is_matinee = True   # try: False, True, False, True
is_member = True    # try: False, False, True, True

# figure out the base price using if / elif / else based on age
if age < 12:
    price = 8.00
elif age >= 65:
    price = 9.00
else:
    price = 12.50

# apply the matinee discount if is_matinee is True
if is_matinee:
    price -= 2.00

# apply the member discount if is_member is True
if is_member:
    price -= 1.00

# print the final price
print("Final price: $%.2f" % price)

# --- Verification against all four table rows ---
def price_for(age, is_matinee, is_member):
    if age < 12:
        p = 8.00
    elif age >= 65:
        p = 9.00
    else:
        p = 12.50
    if is_matinee:
        p -= 2.00
    if is_member:
        p -= 1.00
    return p

test_rows = [(8, False, False), (30, True, False), (70, False, True), (30, True, True)]
for a, m, mem in test_rows:
    print(a, m, mem, "->", "$%.2f" % price_for(a, m, mem))
````
