### Triage

Triage sorts patients by urgency. An `if`, `elif`, `else` chain checks conditions from the top and runs only the first one that is True. Note that `elif` is not a loop. The chain runs once and picks one path.

```python
score = 7
if score >= 9:
    print('High')
elif score >= 5:
    print('Medium')
else:
    print('Low')       # prints: Medium
```

## Part 6: Sets

A set does not allow duplicates, and it has no fixed order. You can compare two sets with `&` (in both) and `|` (in either).

