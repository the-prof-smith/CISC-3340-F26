# Debug the Code: 10 Broken Programs

### Intro to Computing — Debugging Practice

Every code cell below is **broken** on purpose. Your job for each scenario:

1. **Run the cell** and read the error message (or observe the wrong output).
2. **Diagnose** what's wrong — is it a *syntax* error, a *runtime* error, or a *logic* error?
3. **Fix the code** in the cell so it does what the comment/description says it should do.
4. **Re-run** until it works correctly.

> **Tip:** Syntax errors stop the code from running at all. Runtime errors happen while the code is running (like dividing by zero). Logic errors are the sneakiest — the code runs fine but gives you the *wrong answer*.

## Scenario 2: Inconsistent Indentation

**Goal:** This function should return `"even"` or `"odd"` depending on the number.

Run it and look closely at the whitespace in front of each line.


```python
def even_or_odd(n):
    if n % 2 == 0:
        return "even"
      else:
        return "odd"

print(even_or_odd(7))
```
