```python

print(my_tuple[0])  # prints: item one
```

This is where dictionaries come in. A dictionary stores data as key-value pairs. Instead of looking something up by its position like you do in a list or tuple, you look it up by its label, called a key.

```python

    "name": "Harrow Estate",

print(my_dict["floors"]) # prints: 3
```

## Part 3: Sets
### Tracking the Scouted Areas

Dex has been driving past the estate every day this week, logging which areas the crew has already scoped out. A location should never appear twice on the list. Marking the same spot twice wastes time and risks drawing attention.

This is where sets come in. A set is a collection that **does not allow duplicates**. If you try to add something that is already in the set, nothing happens. It just stays the same.

```python
my_set = {"apple", "banana", "cherry"}
my_set.add("apple")  # already there, nothing changes
print(my_set)        # {"apple", "banana", "cherry"}
```

Note: Sets do not have a guaranteed order, so do not be surprised if the items print in a different order each time.
